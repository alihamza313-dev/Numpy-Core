```python
# What is Normalization?

# Normalization means scaling data into a different range without changing the relationship between values.


data = np.array([10, 20, 30, 40, 50])

print(data)

# 1. Min-Max Normalization:
# # fORMULA : 
#     (X - Xmin)/(Xmax - Xmin)

import numpy as np

data = np.array([10, 20, 30, 40, 50])

minimum = np.min(data)
maximum = np.max(data)

print("Minimum:", minimum)
print("Maximum:", maximum)

normalized = (data - minimum) / (maximum - minimum)
print(normalized)
```

```python
# Mean Normalization

# # Formula:
# (X - Mean)/(Xmax - Xmin)

data = np.array([10,20,30,40,50])

mean = np.mean(data)

normalized = (data - mean) / (np.max(data)-np.min(data))

print(normalized)
```

```python
# Z-Score Normalization (Standardization)
# Formula:
#     (x−μ)/σ
# Where
    # μ = mean
    # σ = standard deviation


data = np.array([10,20,30,40,50])

mean = np.mean(data)
std = np.std(data)

print(mean)
print(std)

z = (data - mean) / std

print(z)
```
